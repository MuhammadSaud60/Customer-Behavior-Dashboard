```python
import pandas as pd
```

```python
df = pd.read_csv('customer_shopping_behavior.csv')
```

```python
df[df['Review Rating'].isna()]
```

```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
df.rename(columns={'purchase_amount_(usd)' : 'purchase_amount'}, inplace=True)
```

```python
df.head()
```

```python
# New Column Age Group

labels = ['Young Adult', 'Adult', 'Middle_age', 'Senior']

df['age_group'] = pd.qcut(df['age'], q=4, labels=labels)
```

```python
df.head()
```

```python
df[['age', 'age_group']].head(10)
```

```python
df['frequency_of_purchases'].unique()
```

```python
# New Column Purchase Frequency Days

frequency_mapping = {
    'Fortnightly': 14,
    'Weekly' : 7,
    'Annually': 365,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Monthly': 30,
    'Every 3 Months': 90
}

df['purchace_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['frequency_of_purchases', 'purchace_frequency_days']]
```

```python
df[['discount_applied', 'promo_code_used']].sample(10)
```
